```
# 4. Export the best weights to ONNX for our pipeline (newest run folder)
import glob, os
from ultralytics import YOLO
run = max(glob.glob('runs/*/'), key=os.path.getmtime)
print('Run:', run)
YOLO(run + 'weights/best.pt').export(format='onnx', imgsz=1280, opset=12)
```

```
# 5. Download the results: one zip with best.pt, best.onnx, size_recall.csv and the plots
import shutil
shutil.make_archive('yolo_results', 'zip', run)
files.download('yolo_results.zip')
```
